```python
import pandas as pd
import numpy as np
from scipy.stats import norm

# Load Iris dataset from GitHub
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

# Display first five rows
print("Iris Dataset:")
print(df.head())

# Select a feature
data = df["sepal_length"].values

# Calculate MLE of mean and variance
mle_mean = np.mean(data)
mle_variance = np.var(data, ddof=0)

print("\nMaximum Likelihood Estimation:")
print("Mean (MLE):", mle_mean)
print("Variance (MLE):", mle_variance)
```
